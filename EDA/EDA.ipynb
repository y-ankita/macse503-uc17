```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")

OUTPUT_DIR = "HR_EDA_Visualizations"
os.makedirs(OUTPUT_DIR, exist_ok=True)

def save_plot(filename):
    """Save the current plot and display it."""
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUTPUT_DIR, filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()


def plot_count(df, column, title, filename, top_n=None,
               rotation=45):
    """Plot counts for a categorical column."""
    if column not in df.columns:
        print(f"Skipped: {column} not found")
        return

    data = df[column].dropna()

    if data.empty:
        print(f"Skipped: {column} has no values")
        return

    counts = data.value_counts()

    if top_n:
        counts = counts.head(top_n)

    plt.figure(figsize=(11, 6))
    sns.barplot(x=counts.index.astype(str), y=counts.values)
    plt.title(title)
    plt.xlabel(column)
    plt.ylabel("Number of Records")
    plt.xticks(rotation=rotation, ha="right")
    save_plot(filename)


def plot_histogram(df, column, title, filename, bins=30):
    """Plot a numeric distribution."""
    if column not in df.columns:
        print(f"Skipped: {column} not found")
        return

    values = pd.to_numeric(df[column], errors="coerce").dropna()

    if values.empty:
        print(f"Skipped: {column} has no numeric values")
        return

    plt.figure(figsize=(9, 5))
    sns.histplot(values, bins=bins, kde=True)
    plt.title(title)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    save_plot(filename)


def plot_box(df, column, category, title, filename):
    """Plot a numeric variable across categories."""
    if column not in df.columns or category not in df.columns:
        print(f"Skipped: {column} or {category} not found")
        return

    data = df[[column, category]].copy()
    data[column] = pd.to_numeric(data[column], errors="coerce")
    data = data.dropna()

    if data.empty:
        print(f"Skipped: insufficient data for {title}")
        return

    # Keep the chart readable when there are many categories
    top_categories = data[category].value_counts().head(15).index
    data = data[data[category].isin(top_categories)]

    plt.figure(figsize=(12, 6))
    sns.boxplot(data=data, x=category, y=column)
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    save_plot(filename)


def plot_correlation(df, columns, title, filename):
    """Plot correlations among available numeric columns."""
    available = [c for c in columns if c in df.columns]

    if len(available) < 2:
        print(f"Skipped: not enough columns for {title}")
        return

    data = df[available].copy()

    for col in available:
        data[col] = pd.to_numeric(data[col], errors="coerce")

    data = data.dropna(axis=1, how="all")
    corr = data.corr(numeric_only=True)

    if corr.shape[1] < 2:
        print(f"Skipped: not enough numeric data for {title}")
        return

    plt.figure(figsize=(11, 8))
    sns.heatmap(
        corr,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )
    plt.title(title)
    save_plot(filename)


print("\n========== PERFORMANCE DATASET ==========")

performance["ReviewDate"] = pd.to_datetime(
    performance["ReviewDate"], errors="coerce"
)

performance_numeric = [
    "EnvironmentSatisfaction",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "TrainingOpportunitiesWithinYear",
    "TrainingOpportunitiesTaken",
    "WorkLifeBalance",
    "SelfRating",
    "ManagerRating"
]

for col in performance_numeric:
    if col in performance.columns:
        performance[col] = pd.to_numeric(
            performance[col], errors="coerce"
        )


plot_count(
    performance, "SelfRating",
    "Distribution of Employee Self-Ratings",
    "01_performance_self_rating.png"
)

plot_count(
    performance, "ManagerRating",
    "Distribution of Manager Ratings",
    "02_performance_manager_rating.png"
)

for col, filename in [
    ("EnvironmentSatisfaction", "03_environment_satisfaction.png"),
    ("JobSatisfaction", "04_job_satisfaction.png"),
    ("RelationshipSatisfaction", "05_relationship_satisfaction.png"),
    ("WorkLifeBalance", "06_work_life_balance.png")
]:
    plot_count(
        performance, col,
        f"Distribution of {col}",
        filename
    )


if {"SelfRating", "ManagerRating"}.issubset(performance.columns):
    comparison = performance[
        ["SelfRating", "ManagerRating"]
    ].copy()

    comparison["SelfRating"] = pd.to_numeric(
        comparison["SelfRating"], errors="coerce"
    )
    comparison["ManagerRating"] = pd.to_numeric(
        comparison["ManagerRating"], errors="coerce"
    )

    comparison = comparison.dropna()

    if not comparison.empty:
        rating_comparison = comparison.melt(
            var_name="Rating Type",
            value_name="Rating"
        )

        plt.figure(figsize=(8, 5))
        sns.countplot(
            data=rating_comparison,
            x="Rating",
            hue="Rating Type"
        )
        plt.title("Employee Self-Ratings vs Manager Ratings")
        save_plot("07_self_vs_manager_ratings.png")


training_cols = [
    "TrainingOpportunitiesWithinYear",
    "TrainingOpportunitiesTaken"
]

if all(c in performance.columns for c in training_cols):
    training_data = performance[training_cols].copy()

    for col in training_cols:
        training_data[col] = pd.to_numeric(
            training_data[col], errors="coerce"
        )

    training_data = training_data.dropna()

    if not training_data.empty:
        training_summary = training_data.mean().reset_index()
        training_summary.columns = ["Training Measure", "Average"]

        plt.figure(figsize=(9, 5))
        sns.barplot(
            data=training_summary,
            x="Training Measure",
            y="Average"
        )
        plt.title("Average Training Opportunities: Offered vs Taken")
        plt.xticks(rotation=20, ha="right")
        save_plot("08_training_opportunities.png")


plot_correlation(
    performance,
    performance_numeric,
    "Performance and Satisfaction Correlation Heatmap",
    "09_performance_correlation.png"
)

if "ReviewDate" in performance.columns:
    review_counts = (
        performance.dropna(subset=["ReviewDate"])
        .set_index("ReviewDate")
        .resample("ME")
        .size()
    )

    if not review_counts.empty:
        plt.figure(figsize=(12, 5))
        review_counts.plot()
        plt.title("Number of Performance Reviews Over Time")
        plt.xlabel("Review Month")
        plt.ylabel("Number of Reviews")
        save_plot("10_performance_reviews_over_time.png")




print("\n========== HRMS DATASET ==========")

hrms["hire_date"] = pd.to_datetime(
    hrms["hire_date"], errors="coerce"
)
hrms["termination_date"] = pd.to_datetime(
    hrms["termination_date"], errors="coerce"
)
hrms["last_training_date"] = pd.to_datetime(
    hrms["last_training_date"], errors="coerce"
)
hrms["last_promotion_date"] = pd.to_datetime(
    hrms["last_promotion_date"], errors="coerce"
)

hrms_numeric = [
    "age", "base_salary", "bonus_pct", "equity_grant",
    "equity_pct", "fte", "training_count", "promotion_count"
]

for col in hrms_numeric:
    if col in hrms.columns:
        hrms[col] = pd.to_numeric(hrms[col], errors="coerce")


plot_count(
    hrms, "department",
    "Employee Distribution by Department",
    "11_hrms_department.png",
    top_n=20
)


plot_count(
    hrms, "business_unit",
    "Employees by Business Unit",
    "12_hrms_business_unit.png",
    top_n=20
)


plot_count(
    hrms, "employment_status",
    "Employee Employment Status",
    "13_hrms_employment_status.png"
)


plot_count(
    hrms, "employment_type",
    "Employees by Employment Type",
    "14_hrms_employment_type.png"
)


plot_count(
    hrms, "gender",
    "Employee Gender Distribution",
    "15_hrms_gender.png"
)

plot_histogram(
    hrms, "age",
    "Distribution of Employee Age",
    "16_hrms_age_distribution.png"
)


plot_histogram(
    hrms, "base_salary",
    "Distribution of Base Salary",
    "17_hrms_salary_distribution.png"
)


plot_box(
    hrms, "base_salary", "department",
    "Base Salary by Department",
    "18_hrms_salary_by_department.png"
)


plot_count(
    hrms, "job_level",
    "Employee Distribution by Job Level",
    "19_hrms_job_level.png"
)

plot_count(
    hrms, "location",
    "Employee Distribution by Location",
    "20_hrms_location.png",
    top_n=20
)


if "hire_date" in hrms.columns:
    hire_year = hrms["hire_date"].dropna().dt.year

    if not hire_year.empty:
        hire_counts = hire_year.value_counts().sort_index()

        plt.figure(figsize=(11, 5))
        sns.barplot(
            x=hire_counts.index.astype(str),
            y=hire_counts.values
        )
        plt.title("Number of Employees Hired by Year")
        plt.xlabel("Hire Year")
        plt.ylabel("Employees Hired")
        save_plot("21_hrms_hires_by_year.png")

plot_histogram(
    hrms, "training_count",
    "Distribution of Employee Training Count",
    "22_hrms_training_count.png"
)


plot_histogram(
    hrms, "promotion_count",
    "Distribution of Employee Promotion Count",
    "23_hrms_promotion_count.png"
)


plot_count(
    hrms, "bonus_eligible",
    "Employees by Bonus Eligibility",
    "24_hrms_bonus_eligibility.png"
)


plot_count(
    hrms, "high_potential_flag",
    "High-Potential Employee Flag Distribution",
    "25_hrms_high_potential.png"
)


plot_correlation(
    hrms,
    hrms_numeric,
    "HRMS Numeric Variable Correlation Heatmap",
    "26_hrms_correlation.png"
)


missing_counts = hrms.isna().sum()
missing_counts = missing_counts[missing_counts > 0].sort_values(
    ascending=False
)

if not missing_counts.empty:
    plt.figure(figsize=(12, 7))
    sns.barplot(
        x=missing_counts.values,
        y=missing_counts.index
    )
    plt.title("Missing Values by HRMS Column")
    plt.xlabel("Number of Missing Values")
    plt.ylabel("Column")
    save_plot("27_hrms_missing_values.png")




print("\n========== LEAVE DATASET ==========")

leave_date_cols = ["Start Date", "End Date"]

for col in leave_date_cols:
    if col in leave.columns:
        leave[col] = pd.to_datetime(leave[col], errors="coerce")

leave_numeric = [
    "Days Taken",
    "Total Leave Entitlement",
    "Leave Taken So Far",
    "Remaining Leaves"
]

for col in leave_numeric:
    if col in leave.columns:
        leave[col] = pd.to_numeric(leave[col], errors="coerce")


plot_count(
    leave, "Leave Type",
    "Leave Records by Leave Type",
    "28_leave_type.png"
)


plot_count(
    leave, "Department",
    "Leave Records by Department",
    "29_leave_department.png",
    top_n=20
)


plot_count(
    leave, "Position",
    "Leave Records by Position",
    "30_leave_position.png",
    top_n=15
)


plot_histogram(
    leave, "Days Taken",
    "Distribution of Days Taken",
    "31_leave_days_taken.png"
)


plot_histogram(
    leave, "Remaining Leaves",
    "Distribution of Remaining Leaves",
    "32_leave_remaining.png"
)


plot_histogram(
    leave, "Total Leave Entitlement",
    "Distribution of Total Leave Entitlement",
    "33_leave_entitlement.png"
)

if {"Leave Type", "Days Taken"}.issubset(leave.columns):
    summary = leave.groupby("Leave Type")["Days Taken"].mean()
    summary = summary.dropna().sort_values(ascending=False)

    if not summary.empty:
        plt.figure(figsize=(10, 5))
        sns.barplot(x=summary.index, y=summary.values)
        plt.title("Average Days Taken by Leave Type")
        plt.xlabel("Leave Type")
        plt.ylabel("Average Days Taken")
        plt.xticks(rotation=35, ha="right")
        save_plot("34_leave_average_by_type.png")


leave_scatter_cols = [
    "Total Leave Entitlement", "Leave Taken So Far"
]

if all(c in leave.columns for c in leave_scatter_cols):
    scatter_data = leave[leave_scatter_cols].copy()

    for col in leave_scatter_cols:
        scatter_data[col] = pd.to_numeric(
            scatter_data[col], errors="coerce"
        )

    scatter_data = scatter_data.dropna()

    if not scatter_data.empty:
        plt.figure(figsize=(8, 6))
        sns.scatterplot(
            data=scatter_data,
            x="Total Leave Entitlement",
            y="Leave Taken So Far"
        )
        plt.title("Total Leave Entitlement vs Leave Taken")
        save_plot("35_leave_entitlement_vs_taken.png")


if "month" in leave.columns:
    plot_count(
        leave, "month",
        "Leave Records by Month",
        "36_leave_by_month.png",
        rotation=45
    )


plot_box(
    leave, "Remaining Leaves", "Department",
    "Remaining Leave by Department",
    "37_leave_remaining_by_department.png"
)




print("\n========== ATTENDANCE DATASET ==========")

if "Date" in attendance.columns:
    attendance["Date"] = pd.to_datetime(
        attendance["Date"], errors="coerce"
    )

person_cols = [
    col for col in attendance.columns
    if str(col).startswith("Person_")
]

if person_cols:
    attendance_long = attendance.melt(
        id_vars=["Date"] if "Date" in attendance.columns else [],
        value_vars=person_cols,
        var_name="Employee",
        value_name="RecordedValue"
    )

    attendance_long["RecordedValue"] = (
        attendance_long["RecordedValue"]
        .astype("string")
        .str.strip()
    )

    attendance_long["RecordedValue"] = (
        attendance_long["RecordedValue"]
        .replace(["", "nan", "None", "NaT"], pd.NA)
    )

    
    completeness = (
        attendance_long.assign(
            IsRecorded=attendance_long["RecordedValue"].notna()
        )
        .groupby("Employee")["IsRecorded"]
        .mean()
        .mul(100)
        .sort_values(ascending=False)
    )

    if not completeness.empty:
        plt.figure(figsize=(15, 6))
        sns.barplot(
            x=completeness.index,
            y=completeness.values
        )
        plt.title("Attendance Data Completeness by Employee")
        plt.xlabel("Employee Column")
        plt.ylabel("Recorded Values (%)")
        plt.xticks(rotation=90)
        save_plot("38_attendance_completeness.png")

    
    recorded_counts = (
        attendance_long.groupby("Employee")["RecordedValue"]
        .count()
        .sort_values(ascending=False)
    )

    if not recorded_counts.empty:
        plt.figure(figsize=(15, 6))
        sns.barplot(
            x=recorded_counts.index,
            y=recorded_counts.values
        )
        plt.title("Number of Recorded Attendance Values by Employee")
        plt.xlabel("Employee Column")
        plt.ylabel("Number of Recorded Values")
        plt.xticks(rotation=90)
        save_plot("39_attendance_records_by_employee.png")


    if "Date" in attendance_long.columns:
        daily_counts = (
            attendance_long.dropna(subset=["Date"])
            .groupby("Date")["RecordedValue"]
            .count()
        )

        if not daily_counts.empty:
            plt.figure(figsize=(13, 5))
            daily_counts.plot()
            plt.title("Recorded Attendance Values Over Time")
            plt.xlabel("Date")
            plt.ylabel("Number of Non-Missing Values")
            save_plot("40_attendance_records_over_time.png")

    
    time_values = pd.to_datetime(
        attendance_long["RecordedValue"],
        format="%H:%M:%S",
        errors="coerce"
    )

    
    missing_time = time_values.isna()
    time_values.loc[missing_time] = pd.to_datetime(
        attendance_long.loc[missing_time, "RecordedValue"],
        format="%H:%M",
        errors="coerce"
    )

    valid_times = time_values.dropna()

    if not valid_times.empty:
        time_minutes = (
            valid_times.dt.hour * 60 + valid_times.dt.minute
        )

        plt.figure(figsize=(10, 5))
        sns.histplot(time_minutes, bins=24, kde=False)
        plt.title("Distribution of Recorded Times")
        plt.xlabel("Time of Day (Minutes After Midnight)")
        plt.ylabel("Frequency")
        plt.xticks(
            [0, 240, 480, 720, 960, 1200, 1440],
            ["00:00", "04:00", "08:00", "12:00",
             "16:00", "20:00", "24:00"]
        )
        save_plot("41_attendance_recorded_time_distribution.png")

else:
    print(
        "No Person_* columns were detected. "
        "Inspect attendance.columns and attendance.head() "
        "to determine the XML structure."
    )



print("\n========== ACCESS DATASET ==========")


plot_count(
    access, "target",
    "Distribution of Access Target Labels",
    "42_access_target_distribution.png"
)


plot_count(
    access, "ROLE_TITLE",
    "Most Common Role Titles",
    "43_access_role_title.png",
    top_n=15
)


plot_count(
    access, "ROLE_DEPTNAME",
    "Most Common Role Departments",
    "44_access_role_department.png",
    top_n=15
)


plot_count(
    access, "ROLE_FAMILY",
    "Access Records by Role Family",
    "45_access_role_family.png",
    top_n=15
)


plot_count(
    access, "ROLE_ROLLUP_1",
    "Access Records by Role Rollup 1",
    "46_access_role_rollup_1.png",
    top_n=15
)


plot_count(
    access, "ROLE_ROLLUP_2",
    "Access Records by Role Rollup 2",
    "47_access_role_rollup_2.png",
    top_n=15
)


plot_count(
    access, "RESOURCE",
    "Most Common Resources in Access Records",
    "48_access_resource.png",
    top_n=15
)


access_missing = access.isna().sum()
access_missing = access_missing[access_missing > 0].sort_values(
    ascending=False
)

if not access_missing.empty:
    plt.figure(figsize=(12, 6))
    sns.barplot(
        x=access_missing.values,
        y=access_missing.index
    )
    plt.title("Missing Values by Access Dataset Column")
    plt.xlabel("Missing Values")
    plt.ylabel("Column")
    save_plot("49_access_missing_values.png")



print("\n========== EDA COMPLETE ==========")
print("Visualization folder:", os.path.abspath(OUTPUT_DIR))

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]} columns")

print("\nCharts are saved as PNG files in:", OUTPUT_DIR)
```
